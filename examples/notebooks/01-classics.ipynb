# Bo classics

This notebook adapts `examples/01-classics.bo` for the Bo Jupyter kernel. It introduces recursion, anonymous recursion, loops, and a small primality test.

The Bo kernel already loads `core.bo` and resolves paths from its checkout, so this cell can load the example-only helper words (`say`, `filter`, and `range`), and with them the standard library, regardless of JupyterLab's working directory.

```bo
"examples/lib.bo" require
```

## Recursion

`fact` calls itself by its named word. `fact2` uses `lambda` and `rec` for anonymous recursion.

```bo
    ((dup 0 =) (drop 1) (dup 1 - fact *) ifte)
    (fact)
PUSH

    (((dup 0 =) (drop 1) (dup 1 - rec *) ifte) lambda)
    (fact2)
PUSH

"10! = " emits 10 fact say
"15! = " emits 15 fact say
"10! (rec) = " emits 10 fact2 say
```

## Loops

`loop` consumes a condition quotation and a body quotation. The Fibonacci program carries `(a b n)` on the stack, while `collatz` counts steps before reaching one.

```bo
    (0 1 rot (dup 0 >) (1 - -rot tuck + rot) loop drop drop)
    (fib)
PUSH

    (
    0 swap
        (dup 1 >)
        (dup 2 mod 0 = () (2 //) (3 * 1 +) ifte swap 1 + swap)
    loop drop
    )
    (collatz)
PUSH

"fib(30) = " emits 30 fib say
"gcd(1071, 462) = " emits 1071 462 gcd say
"collatz(27) = " emits 27 collatz say
```

## Primality

`primefrom` tests possible divisors up to the square root of the candidate. `prime?` handles values below two, then begins the search at divisor two.

```bo
    (
        (dup dup * 2 pick >)
        (2drop 1)
        ((over over mod 0 =) (2drop 0) (1 + primefrom) ifte)
    ifte
    )
    (primefrom)
PUSH

    ((dup 2 <) (drop 0) (2 primefrom) ifte)
    (prime?)
PUSH

"2^40 = " emits 2 40 ^ say
"prime(97) = " emits 97 prime? say
"prime(91) = " emits 91 prime? say
"primes below 60: " emits 2 60 range (apply prime?) filter say
```
